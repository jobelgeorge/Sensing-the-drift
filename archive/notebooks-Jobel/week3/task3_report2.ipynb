```python
# Import necessary libraries
import os
import sys
# Add the utils folder to the system path so we can import custom modules
sys.path.append(os.path.abspath(os.path.join("../..")))
import matplotlib.pyplot as plt
import torch
import numpy as np
import torch.nn as nn
import pandas as pd
from utils.reproducibility_utils import seed_everything
from utils.data_loadingV2 import load_data, load_test_data, feature_names

# Set the plotting style for better-looking visualizations
plt.style.available
plt.style.use('seaborn-v0_8')
```

```python
print("Loading train data...")
train_dataloader, val_dataloader, pipeline, mean_dict, std_dict = load_data(
    train_path = "../../datasets_challenge/train",  # Path to training data
    val_file = 'train_part_006.csv',
    train_ratio = 0.8,  # Use 80% for training, 20% for validation
    batchsize = 512,  # Number of samples per batch
    columns_to_standardize = feature_names,  # Option to specify which columns to standardize
    columns_to_drop = ["sideSlip", "vyCG", "time", "longitude","latitude"],  # Remove these from features
    sequence_length = 20,  # Number of time steps in each sequence (t-19 to t0)
    threshold = True,  # Enable velocity thresholding
    threshold_value = 3.0,  # Only keep samples where vehicle speed > 3 m/s
    threshold_column = "vxCG",  # Column to apply threshold on (longitudinal velocity)
    target = "sideSlip",  # The variable we want to predict
    scaling = True,  # Apply standardization to features
    seed = 42,  # Random seed for train/val split
)
```

```python
num_features = len(feature_names)
print("Num features:", num_features)
```

```python
class RNNResidual(nn.Module):
    def __init__(self, input_size, hidden_size=128, num_layers=2, dropout=0):
        super().__init__()
        self.rnn = nn.RNN(input_size, hidden_size, num_layers,
                          nonlinearity='tanh', batch_first=True,
                          dropout=dropout if num_layers > 1 else 0.0)
        self.norm = nn.LayerNorm(hidden_size)
        self.fc = nn.Sequential(
            nn.Linear(hidden_size + input_size, 64),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(64, 1)
        )

    def forward(self, x):
        out, _ = self.rnn(x)
        out = self.norm(out)
        last = out[:, -1, :]
        combined = torch.cat([last, x[:, -1, :]], dim=1)
        return self.fc(combined)

# Define custom loss function: Focal + band-weighted L1 for imbalanced regression
def focal_band_l1(pred, target, gamma=2.0):
    """Focal + band-weighted L1 for imbalanced regression."""
    err = torch.abs(pred - target)
    w = torch.ones_like(target)
    many_mask = torch.abs(target) <= 1.0
    medium_mask = (torch.abs(target) > 1.0) & (torch.abs(target) <= 5.0)
    few_mask = torch.abs(target) > 5.0
    w[many_mask], w[medium_mask], w[few_mask] = 1.5, 2.0, 3.5
    focal = torch.pow(err + 1e-6, gamma)
    return torch.mean(w * focal * err)

def band_mae(y_true, y_pred):
    abs_err = np.abs(y_true - y_pred)
    many = abs_err[np.abs(y_true) <= 1.0].mean()
    medium = abs_err[(np.abs(y_true) > 1.0) & (np.abs(y_true) <= 5.0)].mean()
    few = abs_err[np.abs(y_true) > 5.0].mean()
    return many, medium, few
```

```python
# Set device for computation (CPU or GPU)
device = "cpu"  # Change to "cuda" if you have a GPU available

model = RNNResidual(input_size=num_features, hidden_size=128, dropout=0).to(device)
optimizer = torch.optim.AdamW(model.parameters(), lr=5e-4, weight_decay=1e-5)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=10, eta_min=1e-6)
```

```python
early_stop_thresh = 15
num_epochs = 100

best_loss = np.inf
best_epoch = -1

train_curve, val_curve = [], []

for epoch in range(num_epochs):
    model.train()
    running_loss = 0.0

    #Start plain L1, then switch to focal-band
    if epoch < 10:
        criterion = nn.L1Loss()
        print(f"Epoch {epoch+1}: Using plain L1 loss")
    else:
        criterion = focal_band_l1
        print(f"Epoch {epoch+1}: Using Focal-Band L1 loss")

    # ---- TRAINING ----
    for step, (x, y) in enumerate(train_dataloader):
        X, y = x.to(device), y.to(device).reshape([len(y), 1])

        optimizer.zero_grad()
        pred = model(X)
        train_loss = criterion(pred, y)
        running_loss += train_loss.item()

        train_loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)  # gradient clipping
        optimizer.step()

    avg_train_loss = running_loss / (step + 1)
    train_curve.append(avg_train_loss)

    # ---- VALIDATION ----
    model.eval()
    running_val_loss = 0.0
    y_true_all, y_pred_all = [], []
    with torch.no_grad():
        for val_step, (x, y) in enumerate(val_dataloader):
            X_val, y_val = x.to(device), y.to(device).reshape([len(y), 1])
            preds = model(X_val)
            val_loss = criterion(preds, y_val)
            running_val_loss += val_loss.item()

            y_true_all.append(y_val.cpu().numpy().flatten())
            y_pred_all.append(preds.cpu().numpy().flatten())

    avg_val_loss = running_val_loss / (val_step + 1)
    val_curve.append(avg_val_loss)

    y_true_all = np.concatenate(y_true_all)
    y_pred_all = np.concatenate(y_pred_all)
    mae_many, mae_medium, mae_few = band_mae(y_true_all, y_pred_all)

    # Print progress
    print(f"Epoch {epoch+1:02d}/{num_epochs} | Train Loss: {avg_train_loss:.5f} | "
          f"Val Loss: {avg_val_loss:.5f} | Many: {mae_many:.3f} | Med: {mae_medium:.3f} | Few: {mae_few:.3f}")

    scheduler.step()  # focus scheduler on few-shot performance

    # ---- EARLY STOPPING ----
    if avg_val_loss < best_loss:
        best_loss = avg_val_loss
        best_epoch = epoch
        torch.save(model.state_dict(), "best_rnn_model.pth")
        print(f"  → New best model saved (val loss = {best_loss:.5f})")
    elif epoch - best_epoch > early_stop_thresh:
        print(f"  → Early stopping at epoch {epoch+1} (best epoch: {best_epoch+1})")
        break
```

```python
# Visualize training progress
plt.figure(figsize=(7, 4))
plt.plot(train_curve, label="train")
plt.plot(val_curve, label="val")
plt.axvline(x=best_epoch, color='gray', linestyle='--', label='Best Val Loss')
plt.xlabel("epoch")
plt.ylabel("MAE Loss")
plt.legend()
plt.title("Training and Validation Loss Over Time")
plt.show()
```

```python
def concate_csv_dir(base_dir):
    all_dfs = []
    csv_lengths = {}
    for root, dirs, files in os.walk(base_dir):
        for file in files:
            if file.endswith('.csv'):
                file_path = os.path.join(root, file)
                df = pd.read_csv(file_path)
                relative_path = os.path.relpath(file_path, base_dir)
                csv_lengths[relative_path] = len(df)
                all_dfs.append(df)
    
    return all_dfs, csv_lengths


df_list, csv_lengths = concate_csv_dir("datasets_challenge/train")
train_df = pd.concat(df_list, ignore_index=True)
train_df = train_df.where(train_df["vxCG"] > 3.0).dropna()  # Apply the same thresholding as during training


train_df["band"] = train_df["sideSlip"].round(1).map(shot_band_mapping)

plt.figure(figsize=(20, 6))
plt.scatter(train_df.index, train_df["sideSlip"], c=train_df["band"].map({"many": "blue", "medium": "orange", "few": "red"}), s=1)
# Plot the color code in a legend
plt.legend(handles=[plt.Line2D([0], [0], marker='o', color='w', markerfacecolor='blue', markersize=7, label='many'),
                    plt.Line2D([0], [0], marker='o', color='w', markerfacecolor='orange', markersize=7, label='medium'),
                    plt.Line2D([0], [0], marker='o', color='w', markerfacecolor='red', markersize=7, label='few')],
           title="Shot Bands", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```
