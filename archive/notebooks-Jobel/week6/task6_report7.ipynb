### Learnable Smoothing Layer

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class LearnableSmoothing(nn.Module):
    def __init__(self, kernel_size=5):
        super().__init__()
        self.k = kernel_size

        # This kernel will be learned!
        self.kernel = nn.Parameter(torch.randn(1, 1, kernel_size))

    def forward(self, x):
        # x shape: (batch, 1)
        pad = self.k // 2

        # Pad for convolution
        x_padded = F.pad(x, (pad, pad), mode="reflect")  # reflect gives better boundary behavior
        
        # Normalized kernel (softmax to keep stability)
        kernel_norm = F.softmax(self.kernel, dim=2)

        # Apply convolution = smoothing
        x_conv = F.conv1d(x_padded.unsqueeze(1), kernel_norm)

        return x_conv.squeeze(1)
```

### RNN Model + Learnable Smoothing Layer

```python
class RNNWithLearnableSmooth(nn.Module):
    def __init__(self, input_size, output_size=1,
                 rnn_units=64, fc_units=64,
                 activation_fct=nn.ReLU(),
                 kernel_size=5):
        super().__init__()

        self.rnn_layer = nn.RNN(
            input_size=input_size,
            hidden_size=rnn_units,
            num_layers=1,
            nonlinearity='relu',
            batch_first=True
        )

        self.fc = nn.Linear(rnn_units, fc_units)
        self.output_layer = nn.Linear(fc_units, output_size)

        # ⭐ Learnable smoothing module
        self.smoother = LearnableSmoothing(kernel_size=kernel_size)

        self.activation_fct = activation_fct

    def forward(self, x):
        x, _ = self.rnn_layer(x)
        x = x[:, -1, :]                   # last time step

        x = self.activation_fct(self.fc(x))
        x = self.output_layer(x)          # raw prediction

        # ⭐ Smooth the output with learned kernel
        x = self.smoother(x)

        return x
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")

input_size = X.shape[2]
rnn_units = 64
fc_units = 64
output_size = 1
lr = 0.0003

model = RNNWithLearnableSmooth(
    input_size=input_size,
    output_size=1,
    rnn_units=rnn_units,
    fc_units=fc_units,
    kernel_size=5   # can try 7 or 9
).to(device)

criterion = nn.L1Loss()
optimizer = torch.optim.Adam(model.parameters(), lr=lr)
print(f"Model parameters: {sum(p.numel() for p in model.parameters()):,}")
```

```python
# Training loop
num_epochs = 50
early_stop_thresh = 10
best_loss = np.inf
best_epoch = -1

train_curve = []
val_curve = []

print("\n" + "="*80)
print("🚀 STARTING TRAINING (WITH VALIDATION SPLIT)")
print("="*80)

for epoch in range(num_epochs):
    # TRAINING
    model.train()
    running_loss = 0
    
    for step, (x, y) in enumerate(train_dataloader):
        X, y = x.to(device), y.to(device).reshape([len(y), 1])
        
        optimizer.zero_grad()
        pred = model(X)
        train_loss = criterion(pred, y)
        running_loss += train_loss.item()
        train_loss.backward()
        optimizer.step()
    
    avg_train_loss = running_loss / (step + 1)
    train_curve.append(avg_train_loss)

    # VALIDATION
    if val_dataloader is not None:
        model.eval()
        running_val_loss = 0
        with torch.no_grad():
            for val_step, (x, y) in enumerate(val_dataloader):
                X_val, y_val = x.to(device), y.to(device).reshape([len(y), 1])
                val_loss = criterion(model(X_val), y_val)
                running_val_loss += val_loss.item()
        
        avg_val_loss = running_val_loss / (val_step + 1)
        val_curve.append(avg_val_loss)

        print(f"Epoch {epoch+1:2d}/{num_epochs} | Train: {avg_train_loss:.5f} | Val: {avg_val_loss:.5f}", end="")
        
        # EARLY STOPPING
        if avg_val_loss < best_loss:
            best_loss = avg_val_loss
            best_epoch = epoch
            # 🔥 FIX: Complete save dictionary
            torch.save({
                'epoch': epoch,
                'model_state_dict': model.state_dict(),
                'optimizer_state_dict': optimizer.state_dict(),
                'loss': best_loss,
                'train_loss': avg_train_loss,
                'val_loss': avg_val_loss
            }, 'best_model.pth')
            print(f" 💾 NEW BEST!")
        elif epoch - best_epoch > early_stop_thresh:
            print(f"\n⏹️  Early stopping at epoch {epoch+1}")
            break
        else:
            print("")
    else:
        print(f"⚠️  WARNING: No validation split! Training on 100%")
        print(f"Epoch {epoch+1}/{num_epochs} | Train: {avg_train_loss:.5f}")

# LOAD BEST MODEL
if best_epoch != -1:
    checkpoint = torch.load('best_model.pth', weights_only=True)
    model.load_state_dict(checkpoint['model_state_dict'])
    print(f"\n" + "="*80)
    print(f"✅ Best model from epoch {best_epoch+1} loaded")
    print(f"   Train Loss: {checkpoint.get('train_loss', 'N/A'):.5f}")
    print(f"   Val Loss:   {checkpoint.get('val_loss', best_loss):.5f}")
    print("="*80)
else:
    print("\n⚠️  No improvement found, using current model")
```

```python
# Plot training curves
plt.figure(figsize=(10, 5))
plt.plot(train_curve, label='Train Loss', linewidth=2)
if val_curve:
    plt.plot(val_curve, label='Val Loss', linewidth=2)
    if best_epoch != -1:
        plt.axvline(x=best_epoch, color='r', linestyle='--', alpha=0.5, label=f'Best Epoch ({best_epoch+1})')
plt.xlabel('Epoch')
plt.ylabel('MAE Loss')
plt.title('Training Progress')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print(f"Final train loss: {train_curve[-1]:.5f}")
if val_curve:
    print(f"Final val loss:   {val_curve[-1]:.5f}")
    print(f"Best val loss:    {best_loss:.5f} (epoch {best_epoch+1})")
```
