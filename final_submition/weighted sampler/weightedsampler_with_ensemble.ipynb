```python
import sys
from pathlib import Path
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F
import plotly.graph_objects as go
import warnings
import copy
warnings.filterwarnings('ignore')

# Path
PROJECT_ROOT = Path.cwd().parents[1]
sys.path.insert(0, str(PROJECT_ROOT))
from utils.reproducibility_utils import seed_everything
from utils.data_loadingbins import load_data, load_test_data
from models.vanilla_rnn import VanillaRNN
from sklearn.metrics import mean_absolute_error

 # Load data
TRAIN_PATH = PROJECT_ROOT / "datasets_challenge" / "train"
TEST_PATH = PROJECT_ROOT / "datasets_challenge" / "test"
DIST_CSV = PROJECT_ROOT / "datasets_challenge" / "sideSlip_distribution.csv"
MODEL_DIR = PROJECT_ROOT / "expert_models"

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
seed_everything(42) # Use your existing reproducibility util
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"

model = VanillaRNN(
    input_size=22,          # number of features
    hidden_size=48,
    num_layers=1,
    dropout=0.0
).to(device)

criterion = nn.L1Loss()     # MAE (matches few-shot objective)
optimizer = torch.optim.Adam(model.parameters(), lr=3e-4)
```

```python

def train_one_epoch(model, dataloader, optimizer, criterion, device):
    model.train()
    total_loss = 0.0

    for X, y in dataloader:
        X = X.to(device)
        y = y.to(device).unsqueeze(1)

        optimizer.zero_grad()
        preds = model(X)
        loss = criterion(preds, y)
        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    return total_loss / len(dataloader)
```

```python
def validate(model, dataloader, device):
    model.eval()
    preds, targets = [], []

    with torch.no_grad():
        for X, y in dataloader:
            X = X.to(device)
            y = y.to(device)

            pred = model(X).squeeze(1)
            preds.append(pred.cpu().numpy())
            targets.append(y.cpu().numpy())

    preds = np.concatenate(preds)
    targets = np.concatenate(targets)

    return mean_absolute_error(targets, preds)
```

```python
print("Loading train data...")
train_loader, val_loader, pipeline = load_data(
            train_path=TRAIN_PATH,
            dist_csv_path=DIST_CSV,
            val_file_name="train_part_017.csv", 
            columns_to_drop=["sideSlip", "vyCG", "time", "longitude","latitude"],
            sequence_length=20,
            threshold=True,
            threshold_value=3.0,
            threshold_column="vxCG",
            target='sideSlip',
            batchsize=64
        )
```

```python
EPOCHS = 50
SNAPSHOT_EPOCHS = [5, 10, 15, 20, 25, 30, 35, 40, 45, 50]   # good default
snapshots = []
best_val_mae = float("inf")
best_model_state = None

for epoch in range(EPOCHS):
    # -------- TRAIN --------
    train_loss = train_one_epoch(
        model,
        train_loader,
        optimizer,
        criterion,
        device
    )

    # -------- VALIDATE --------
    val_mae = validate(
        model,
        val_loader,
        device
    )

    print(
        f"Epoch {epoch+1:02d} | "
        f"Train Loss: {train_loss:.4f} | "
        f"Val MAE: {val_mae:.4f}"
    )

    # -------- SAVE BEST MODEL --------
    if val_mae < best_val_mae:
        best_val_mae = val_mae
        best_model_state = copy.deepcopy(model.state_dict())

    # save snapshots
    if (epoch + 1) in SNAPSHOT_EPOCHS:
        snapshots.append(copy.deepcopy(model.state_dict()))
        print(f"✓ Snapshot saved at epoch {epoch+1}")
        
# load best model (optional but clean)
model.load_state_dict(best_model_state)

# collect targets once
targets = []
with torch.no_grad():
    for _, y in val_loader:
        targets.append(y.numpy())
targets = np.concatenate(targets)

# snapshot predictions
snapshot_preds = []
for i, state in enumerate(snapshots):
    snap_model = copy.deepcopy(model)
    snap_model.load_state_dict(state)
    snap_model.to(device)
    snap_model.eval()

    preds = []
    with torch.no_grad():
        for X, _ in val_loader:
            X = X.to(device)
            pred = snap_model(X).squeeze(1)
            preds.append(pred.cpu().numpy())

    snapshot_preds.append(np.concatenate(preds))

ensemble_preds = np.mean(snapshot_preds, axis=0)
ensemble_mae = mean_absolute_error(targets, ensemble_preds)
print(f"✓ Ensemble validation MAE: {ensemble_mae:.4f}")
```

```python
train_dataloader_unshuffled = load_test_data(
    test_path=TRAIN_PATH,
    batchsize=64,
    pipeline=pipeline,  # Use the same preprocessing as training
    load_anonymized_data=False  # We have ground truth for training data
)
```

```python
import plotly.graph_objects as go

# Set model to evaluation mode
model.eval()

# Storage for predictions and ground truth
total_loss = 0  
preds = []
ground_truths = []

# Generate predictions for all training data
with torch.no_grad():
    for step, (x, y) in enumerate(train_dataloader_unshuffled):
        X, y = x.to(device), y.to(device).reshape([len(y), 1])

        pred = model(X)
        loss = criterion(pred, y)
        total_loss += loss.item()
        
        # Store predictions and ground truth
        preds.append(pred.reshape(-1).cpu().numpy())
        ground_truths.append(y.reshape(-1).cpu().numpy())

    print(f"Avg. Loss: {total_loss/step}")
    
    # Concatenate all batches into single arrays
    ground_truths = np.concatenate(ground_truths)
    preds = np.concatenate(preds)

# Create interactive plot
fig = go.Figure()
fig.add_trace(go.Scatter(y=ground_truths, mode='lines', name='truth', line=dict(color='blue')))
fig.add_trace(go.Scatter(y=preds.flatten(), mode='lines', name='pred', line=dict(color='red'), opacity=0.8))
fig.update_layout(
    title='Ground Truth vs Predictions',
    xaxis_title='Sample Index',
    yaxis_title='Sideslip Angle (degrees)',
    width=1200,
    height=500
)
fig.show()
```
